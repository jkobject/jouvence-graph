```python
import os

if os.environ.get("JOUVENCE_LIVE_GCS") != "1":
    live_result = {"status": "SKIPPED", "reason": "set JOUVENCE_LIVE_GCS=1 and caller-owned JOUVENCE_BILLING_PROJECT"}
else:
    import gcsfs
    import pyarrow.parquet as pq

    billing_project = os.environ.get("JOUVENCE_BILLING_PROJECT")
    if not billing_project:
        raise RuntimeError("JOUVENCE_BILLING_PROJECT is required for requester-pays live verification")
    fs = gcsfs.GCSFileSystem(project=billing_project, requester_pays=billing_project, version_aware=True)
    if not fs.version_aware:
        raise RuntimeError("version-aware GCS reads are required")
    selected_id = os.environ.get("JOUVENCE_OUTPUT_ID", RECORDS[0]["layer"] + "__" + RECORDS[0]["name"])
    selected = [record for record in RECORDS if record["layer"] + "__" + record["name"] == selected_id]
    if len(selected) != 1:
        raise RuntimeError("JOUVENCE_OUTPUT_ID must select exactly one output owned by this notebook")
    RECORD = selected[0]
    object_path = RECORD["canonical_uri"].removeprefix("gs://")
    if "#" in object_path:
        raise RuntimeError("canonical object key must not already contain a generation suffix")
    expected_generation = RECORD["qc"]["generation"]
    pinned_path = object_path + "#" + expected_generation
    info = fs.info(pinned_path)
    if str(info.get("generation", "")) != expected_generation:
        raise RuntimeError("generation-qualified object resolved a different generation")
    with fs.open(pinned_path, "rb") as handle:
        parquet = pq.ParquetFile(handle)
        live_fields = [
            {"name": field.name, "dtype": str(field.type), "nullable": bool(field.nullable)}
            for field in parquet.schema_arrow
        ]
        live_rows = parquet.metadata.num_rows
        live_row_groups = parquet.metadata.num_row_groups
    if live_rows != RECORD["qc"]["rows"]:
        raise RuntimeError("live footer row count differs from the frozen registry")
    if live_row_groups != RECORD["qc"]["row_groups"]:
        raise RuntimeError("live footer row-group count differs from the frozen registry")
    if live_fields != RECORD["fields"]:
        raise RuntimeError("live footer schema differs from the frozen registry")
    live_result = {"status": "PASS", "generation": str(info["generation"]), "rows": live_rows, "row_groups": live_row_groups}
live_result
```
